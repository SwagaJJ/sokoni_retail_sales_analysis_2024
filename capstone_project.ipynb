## **3. Data Cleaning**

The data set contains only missing values in 3 column sections:
* customer_satisfaction    119
* profit                    12
* total_sales                6

The unique values of the data set are consistent.

* total_sales can be calculated using a formula of unit_price * quantity
* profit to be calculated from the (cost_price-unit_price) * quantity
* customer_satisfaction with no entries to be regarded as "No rating" but retained as NaN for purposes of analysis
* Date format to be converted to DateTime.

### **3.1 Filling total-sales**

```python
# Record which values were originally missing
df["Total_sales_WasMissing"] = df["total_sales"].isna()

# Start cleaning by filling the missing values of the total sales
df["total_sales"] = df["total_sales"].fillna(
    df["quantity"] * df["unit_price"]
)
display(df['total_sales'].value_counts())

# check if all the entries have been filled
print("Missing total_sales after cleaning:", df["total_sales"].isna().sum())
```

### **3.2 Filling profit**

```python
# Record which values were originally missing
df["Profit_WasMissing"] = df["profit"].isna()

# Fill the missing values of the profit
df["profit"] = df["profit"].fillna(
    df["quantity"] *(df["unit_price"] - df["cost_price"])
)
display(df['profit'].value_counts())

# check if all the entries have been filled
print("Missing profit after cleaning:", df["profit"].isna().sum())
```

### **3.3 Changing date format**

```python
# Convert date to datetime object
df["date"] = pd.to_datetime(df["date"])

df.info()
```

## **5. Check the validity of the features of the values**

```python
# Check total_sales calculation consistency
sales_diff = (df["total_sales"] - (df["quantity"] * df["unit_price"])).abs()
print(
    f"Mismatched sales entries: {(sales_diff > 0.01).sum()}"
)  
# Should return 0
```

```python
# Check profit calculation consistency
profit_diff = (
    df["profit"] - (df["quantity"] * (df["unit_price"] - df["cost_price"]))
).abs()
print(
    f"Mismatched profit entries: {(profit_diff > 0.01).sum()}"
)  # Should return 0
```

## **6. Visualization**

```python
#Branch Sales Analysis
branch_sales = (
    df.groupby("branch")["total_sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

print("=== Q1: Total Sales by Branch ===")
display(branch_sales)
# Q1 Visualization: Bar Chart
plt.figure(figsize=(9, 5))
bars1 = plt.bar(
    branch_sales["branch"], branch_sales["total_sales"], color="#2b5c8f"
)

plt.title("Total Sales by Branch", fontsize=13, weight="bold")
plt.xlabel("Branch", fontsize=10)
plt.ylabel("Total Sales (KES)", fontsize=10)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

```python
# Q2. Electronics Profitability per Order

# Filter for Electronics category
electronics_df = df[df["product_category"] == "Electronics"].copy()

segment_col = (
    "customer_segment" if "customer_segment" in df.columns else "payment_method"
)

electronics_profit = (
    electronics_df.groupby(segment_col)["profit"]
    .agg(["mean", "count", "sum"])
    .reset_index()
    .rename(
        columns={
            "mean": "avg_profit_per_order",
            "count": "order_count",
            "sum": "total_profit",
        }
    )
    .sort_values(by="avg_profit_per_order", ascending=False)
)

print(f"\n=== Q2: Electronics Average Profit per Order (by {segment_col}) ===")
display(electronics_profit)
```

```python
# Q2 Visualization: Bar Chart
plt.figure(figsize=(8, 5))
bars2 = plt.bar(
    electronics_profit[segment_col],
    electronics_profit["avg_profit_per_order"],
    color="#2ca02c",
)

plt.title(
    f"Electronics: Average Profit per Order by {segment_col.replace('_', ' ').title()}",
    fontsize=13,
    weight="bold",
)
plt.xlabel(segment_col.replace("_", " ").title(), fontsize=10)
plt.ylabel("Average Profit per Order (KES)", fontsize=10)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

```python
# Q3. 12-Month Profit Trend (2024)

# Group profit by month
monthly_profit = (
    df.groupby(["month_num", "month"])["profit"]
    .sum()
    .reset_index()
    .sort_values("month_num")
)

print("\n=== Q3: Monthly Profit Breakdown ===")
display(monthly_profit[["month", "profit"]])
```

```python
# Q3 Visualization: Line Chart + Trend Line
plt.figure(figsize=(11, 5))
x_numeric = np.arange(len(monthly_profit))
y_profit = monthly_profit["profit"].values

# Plot monthly profit line
plt.plot(
    monthly_profit["month"],
    y_profit,
    marker="o",
    linewidth=2,
    color="#1f77b4",
    label="Monthly Profit",
)

plt.title("2024 Monthly Profit Trend with Trend Line", fontsize=13, weight="bold")
plt.xlabel("Month", fontsize=10)
plt.ylabel("Total Profit (KES)", fontsize=10)
plt.xticks(rotation=45)
plt.legend(fontsize=10)
plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
```
